```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt 
import seaborn as sns

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df
```

```python
country = 'United States'
df_country = df[df['job_country']== country].copy()
df_skills = df_country.explode('job_skills')

```

```python
df_skills
```

```python
df_skill_count = df_skills.groupby(['job_skills', 'job_title_short']).size()
df_skill_count = df_skill_count.reset_index(name='skill_count')
df_skill_count.sort_values(by='skill_count', ascending=False, inplace=True)
df_skill_count
```

```python
roles = ['Data Analyst', 'Data Scientist', 'Data Engineer']
fig, ax = plt.subplots(3,1)
for i, role in enumerate(roles):
    df_plot = df_skill_count[df_skill_count['job_title_short'] == role].head(5)
    df_plot.plot(kind='barh', x='job_skills', y='skill_count', ax=ax[i], legend=False)
    ax[i].invert_yaxis()
    ax[i].set_ylabel('')
    ax[i].set_title(role)

fig.tight_layout()
```

```python
df_job_total = df_country['job_title_short'].value_counts().reset_index(name='jobs_total')
df_job_total
```

```python
df_skill_perc = pd.merge(df_skill_count, df_job_total, how='left', on='job_title_short')

df_skill_perc['skill_perc'] = 100 * df_skill_perc['skill_count']/df_skill_perc['jobs_total']
df_skill_perc
```

```python
roles = ['Data Analyst', 'Data Scientist', 'Data Engineer']
fig, ax = plt.subplots(3,1)
for i, role in enumerate(roles):
    df_plot = df_skill_perc[df_skill_perc['job_title_short'] == role].head(5)
    sns.barplot(data=df_plot, x='skill_perc', y='job_skills', ax=ax[i], hue='skill_perc', palette='dark:r_r')
    ax[i].get_legend().remove()
    
    ax[i].set_ylabel('')
    ax[i].set_xlabel('')
    ax[i].set_title(role)
    ax[i].set_xlim(0,78)
   

    for n,v in enumerate(df_plot['skill_perc']): #n is index and v is value
        ax[i].text(v +1,n, f'{v:.0f}%', va='center')
    if i != len(roles) -1:
        ax[i].set_xticks([])

    

fig.suptitle(f'Likelihood of Skill Requested in {country} Job Postings', fontsize=14)
fig.tight_layout()
```
